### Problem A: Reproducible Matrix Normalization

**Approach:**
We set a random seed using `np.random.seed(2112)` to guarantee reproducible pseudo-random matrix generation. Using a 5x5 random array $X$, we calculate the overall mean ($\mu$) and standard deviation ($\sigma$). The array is normalized via standard score transformation $Z = \frac{X - \mu}{\sigma}$ and exported as a binary `.npy` file.

```python
import numpy as np

# Set seed for reproducible random generation
np.random.seed(2112)

# Generate 5x5 array of uniform random floats in [0, 1)
X = np.random.rand(5, 5)

# Calculate matrix statistics
mean = X.mean()
std = X.std()

# Apply standard score normalization
X_normalized = (X - mean) / std

# Export normalized array to binary format
np.save("X_normalized.npy", X_normalized)

# --- Output Verification ---
print("Mean:", mean)
print("Standard Deviation:", std)
print("\nNormalized Matrix:\n", X_normalized)
```

### Problem B: Cubes Divisible by 4

**Approach:**
We generate an array of the first 100 positive integers ($1$ to $100$) using `np.arange`, reshape it into a 10x10 matrix, and compute the element-wise cube ($x^3$). A Boolean mask evaluates elements where $x^3 \pmod 4 == 0$ to isolate divisible terms into a 1D array before saving to disk.

```python
# Generate 10x10 grid of integers 1 through 100
arr = np.arange(1, 101).reshape(10, 10)

# Vectorized element-wise cubing
cubed_arr = arr ** 3

# Boolean filtering for divisibility by 4
div_by_4 = cubed_arr[cubed_arr % 4 == 0]

# Export filtered array
np.save("div_by_4.npy", div_by_4)

# --- Output Verification ---
print("Shape of filtered array:", div_by_4.shape)
print("First 10 elements divisible by 4:\n", div_by_4[:10])
```

### Problem C: Above-Mean Squares

**Approach:**
We create a 6x6 matrix of squared positive integers ($1^2$ to $36^2$). After computing the scalar mean of the matrix, relational comparison (`> mean`) extracts all elements strictly greater than the overall average. The filtered array is saved as a binary output.

```python
# Create 6x6 matrix of consecutive integers squared
arr_seq = np.arange(1, 37).reshape(6, 6)
squared_arr = arr_seq ** 2

# Compute scalar mean of all elements
matrix_mean = squared_arr.mean()

# Relational Boolean masking
above_mean = squared_arr[squared_arr > matrix_mean]

# Export result to binary format
np.save("above_mean.npy", above_mean)

# --- Output Verification ---
print("Matrix Mean:", matrix_mean)
print("Elements above mean:\n", above_mean)
```

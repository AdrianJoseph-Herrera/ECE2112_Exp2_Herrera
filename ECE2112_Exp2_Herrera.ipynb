### Problem A: Reproducible Matrix Normalization

**Approach:**
We set a random seed using `np.random.seed(2112)` to guarantee reproducible pseudo-random matrix generation. Using a 5x5 random array $X$, we calculate the overall mean ($\mu$) and standard deviation ($\sigma$). The array is normalized via standard score transformation $Z = \frac{X - \mu}{\sigma}$ and exported as a binary `.npy` file.

```python
import numpy as np

# Set seed for reproducible random generation
np.random.seed(2112)

# Generate 5x5 array of uniform random floats in [0, 1)
X = np.random.rand(5, 5)

# Calculate matrix statistics
mean = X.mean()
std = X.std()

# Apply standard score normalization
X_normalized = (X - mean) / std

# Export normalized array to binary format
np.save("X_normalized.npy", X_normalized)

# --- Output Verification ---
print("Mean:", mean)
print("Standard Deviation:", std)
print("\nNormalized Matrix:\n", X_normalized)
```
